```python
import pandas as pd
```

```python
NED_DATASET_PATH = "./datasets/ned_dataset.csv"
SCARA_DATASET_PATH = "./datasets/scara_dataset.csv"
```

```python
ned_data = pd.read_csv(NED_DATASET_PATH)
scara_data = pd.read_csv(SCARA_DATASET_PATH)
```

```python
ned_data.head()
```

```python
scara_data.head()
```

```python
ned_data.describe()
```

```python
scara_data.describe()
```

## Combine both NED and SCARA datasets

```python
combined_data = pd.concat([ned_data, scara_data], ignore_index=True)
```

```python
combined_data.describe()
```

```python
combined_data.info()
```

## Shuffle combined dataset

```python
shuffled_data = combined_data.sample(frac=1, random_state=42, ignore_index=True)
```

```python
shuffled_data.head()
```

## Export shuffled dataset into CSV format

```python
DATASET_FILE_NAME = "./datasets/combined_dataset.csv"
```

```python
shuffled_data.to_csv(DATASET_FILE_NAME, index=False)
```
