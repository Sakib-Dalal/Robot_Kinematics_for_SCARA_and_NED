```python
import matplotlib.pyplot as plt

# Features and their corresponding standard deviation values
features = ['is_scara', 'is_ned', 'q1', 'q2', 'q3', 'x', 'y', 'z']
std_values = [0.500005, 0.500005, 1.138676, 0.413765, 0.465358, 0.298095, 0.190091, 0.215050]

# Set up the figure
plt.figure(figsize=(10, 6))

# Create the bar chart
bars = plt.bar(features, std_values, color='steelblue', edgecolor='black')

# Add titles and labels
plt.title('Standard Deviation of Dataset Features', fontsize=14, fontweight='bold')
plt.xlabel('Features', fontsize=12)
plt.ylabel('Standard Deviation (std)', fontsize=12)

# Annotate each bar with its exact value
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.01, 
             f'{yval:.4f}', ha='center', va='bottom', fontsize=10)

# Add a subtle grid for easier readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Adjust layout and display the plot
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Features and their corresponding mean values
features = ['is_scara', 'is_ned', 'q1', 'q2', 'q3', 'x', 'y', 'z']
mean_values = [0.500000, 0.500000, -0.000859, -0.076387, -0.188522, 0.275374, 0.010745, 0.300515]

# Set up the figure
plt.figure(figsize=(10, 6))

# Create the bar chart
bars = plt.bar(features, mean_values, color='coral', edgecolor='black')

# Add titles and labels
plt.title('Mean Values of Dataset Features', fontsize=14, fontweight='bold')
plt.xlabel('Features', fontsize=12)
plt.ylabel('Mean', fontsize=12)

# Add a horizontal line at y=0 for reference (since some means are negative)
plt.axhline(0, color='black', linewidth=0.8)

# Annotate each bar with its exact value
for bar in bars:
    yval = bar.get_height()
    # Adjust vertical alignment and offset based on whether the value is positive or negative
    offset = 0.01 if yval >= 0 else -0.01
    va = 'bottom' if yval >= 0 else 'top'
    plt.text(bar.get_x() + bar.get_width()/2, yval + offset, 
             f'{yval:.4f}', ha='center', va=va, fontsize=10)

# Add a subtle grid
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Adjust layout and display the plot
plt.tight_layout()
plt.show()
```
